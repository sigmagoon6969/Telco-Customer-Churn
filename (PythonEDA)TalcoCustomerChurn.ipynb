```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.head()
```

```python
df=df.rename(columns={"customerid":"customer_id"})
```

```python
df.head()
```

```python
df["total_charges"]=pd.to_numeric(df["total_charges"],errors="coerce")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["total_charges"]=df["total_charges"].fillna(0.0)
```

```python
df["customer_id"].duplicated().sum()
```

```python
df["multiple_lines"]=df["multiple_lines"].replace("No phone service","No")
df["online_security"]=df["online_security"].replace("No internet service","No")
df["online_backup"]=df["online_backup"].replace("No internet service","No")
df["device_protection"]=df["device_protection"].replace("No internet service","No")
df["tech_support"]=df["tech_support"].replace("No internet service","No")
df["streaming_tv"]=df["streaming_tv"].replace("No internet service","No")
df["streaming_movies"]=df["streaming_movies"].replace("No internet service","No")
df["churn_reason"]=df['churn_reason'].fillna("Active Customer")
```

```python
bins=[-1,12,24,36,48,60,100]
labels=["0-1 Years","1-2 Years","2-3 Years","3-4 Years","4-5 Years","5+ Years"]
df["tenure_group"]=pd.cut(df["tenure_months"],bins=bins,labels=labels)
```

```python
service_columns=["phone_service","multiple_lines",
                 "online_security","online_backup",
                 "device_protection","tech_support",
                 "streaming_tv","streaming_movies"
                 ]
df["services_taken"]=(df[service_columns]=="Yes").sum(axis=1)
```

```python
df=df.drop(columns="lat_long")
```
